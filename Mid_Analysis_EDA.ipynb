```python
import numpy as np
import pandas as pd
import plotly_express as px
```

```python
df = pd.read_csv("C:\\Users\\nesre\\Desktop\\Projects\\KaggleV2-May-2016.csv")
df.head()
```

```python
df.head()
df.shape
df.info()
df.describe()
df.isnull().sum()
df.duplicated().sum()
```

```python
df.columns = df.columns.str.strip().str.lower().str.replace('-', '_')

df.rename(columns={'hipertension': 'hypertension'}, inplace=True)

df['scheduledday'] = pd.to_datetime(df['scheduledday'])
df['appointmentday'] = pd.to_datetime(df['appointmentday'])

df = df[df['age'] >= 0]

df['waiting_days'] = (df['appointmentday'] - df['scheduledday']).dt.days

df = df[df['waiting_days'] >= 0]

df.drop_duplicates(inplace=True)

df.shape
df.isnull().sum()
```

```python
df['no_show'] = df['no_show'].map({'Yes': 1, 'No': 0})
df['no_show'].value_counts()
```

```python
fig = px.histogram(df, x='no_show',title="Show (0) vs No-show (1)")
fig.show()
```

```python
fig = px.box(df, x='no_show', y='age',title="Age vs No-show")
fig.show()
```

```python
fig = px.histogram(df, x='sms_received', color='no_show',barmode='group',title="SMS Received vs No-show")
fig.show()
```

```python
fig = px.histogram(df, x='gender', color='no_show',barmode='group',title="Gender vs No-show")
fig.show()
```

```python
fig = px.histogram(df, x='hypertension', color='no_show',barmode='group',title="Hypertension vs No-show")
fig.show()
```

```python
fig = px.histogram(df, x='waiting_days', nbins=30,title="Waiting Days Distribution")
fig.show()
```
